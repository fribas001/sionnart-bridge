# Run the packaged scenes with native Sionna RT
Use a Sionna environment matching requirements-simulation.txt, plus Jupyter. This notebook defaults to input inspection; switch RUN_SIMULATIONS to True only when ready. The standalone runner imports neither Blender nor add-on modules.

```python
from pathlib import Path
import sys, json
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'metadata/studies.json').is_file())
sys.path.insert(0, str(ROOT / 'scripts'))
print('Package:', ROOT.name)
```

```python
RUN_SIMULATIONS = False
STUDY = 'foliage_r1'
FRAMES = '1,9,10'
records = json.loads((ROOT / 'data' / STUDY / 'configurations.json').read_text())
print('Available frames:', [r['frame'] for r in records])
print('First-frame solver:', records[0]['simulation'])
print('Original/reconstructed:', records[0]['geometry_status'])
```

```python
import subprocess, datetime
if RUN_SIMULATIONS:
    stamp = datetime.datetime.now().strftime('%Y%m%d_%H%M%S')
    output = ROOT.parent / ('native_notebook_' + stamp)
    subprocess.run([sys.executable, str(ROOT / 'scripts/run_study.py'), STUDY, '--frames', FRAMES, '--output', str(output)], check=True)
    print('New results:', output)
else:
    print('Simulation disabled. Change RUN_SIMULATIONS to True to calculate.')
```
